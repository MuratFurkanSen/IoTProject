```python
import asyncio

async def predict_loop():
    while True:
        print("🔮 Predicting...")
        await asyncio.sleep(1)

async def online_training():
    while True:
        await asyncio.sleep(10)   # every 10s (30 min in real life)
        print("🧠 Online training started")
        await asyncio.sleep(3)    # simulate training
        print("✅ Online training done")

async def main():
    asyncio.create_task(online_training())
    await predict_loop()          # never blocked

asyncio.run(main())
```
